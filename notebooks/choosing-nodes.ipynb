```python
import glob
import pickle
import jax.numpy as jnp
import matplotlib.pyplot as plt
```

```python
for filename in glob.glob('../svi_results/full_mm_*'):
    with open(filename, 'rb') as input_file_:
        full_dict = pickle.load(input_file_)
        med_loss = jnp.median(full_dict['svi_results'].losses[-2000:])
        if med_loss < 1e12:
            print(filename)
            print(med_loss)
            # plt.plot(full_dict['svi_results'].losses)
            # plt.ylim(726650, 728000)
            # plt.show()
            print('')
```
